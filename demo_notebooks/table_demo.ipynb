### Imports

```python
import pandas as pd
import numpy as np
import os, sys

if "../" not in sys.path:
    sys.path.append("../")
from hmp_utils import Table
```

### Load a csv to get the data

```python
# the path of the csv to load
csv_path = "./example_csv/input.csv"
# open it as a pandas dataframe
dataframe = pd.read_csv(csv_path)
# convert the dataframe as a numpy array
np_arr = np.array(dataframe)
# in this csv the first column contains the names of the rows, so let's put them in index
index = list(np_arr[:, 0])
# remove the index from the array
np_arr = np_arr[:, 1:]
# the names of the columns can be obtained from the dataframe
header = list(dataframe.columns)
# remove the first value of header because it is the name of the index column (empty in this case)
header = header[1:]
# choose a name for the table
table_name = "Euler Error"
```

### The header and index are the following

```python
print(f"index  = {index}")
print(f"header = {header}")
```

### Use of Table

```python
print("\nTABLE 1:")
table1 = Table(np_arr, table_name=table_name, header=header, index=index)
# you can specifty which rows and columns should be printed
table1.print(rows=[0, 4, 13], columns=list(range(9)))
table1.print_ms(ms_to_print=[80, 160, 320, 400, 560, 1000], framerate=25)

table2 = Table(np_arr[:7, :7], table_name="Output")
# with to_string=True the table will be converted to a string
string_table2 = table2.print(to_string=True)
print("\nTABLE 2:")
print(string_table2)

# you can save the table as a csv
table2.to_csv(path="./example_csv",file_name="output")
```
