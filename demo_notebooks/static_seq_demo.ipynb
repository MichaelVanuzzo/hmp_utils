```python
from hmp_utils import MocapLoader, SmplConstants, ForwardKinematics, plot_static_seq
import numpy as np

mocapLoader = MocapLoader(SmplConstants.BODY_MODEL_TYPE, SmplConstants.KEYPOINTS, 25)
forwardKinematics = ForwardKinematics(SmplConstants.OFFSETS, SmplConstants.PARENTS)
forwardKinematics.set_body_model_type(SmplConstants.BODY_MODEL_TYPE)

FRAMES_TO_VISUALIZE = [2, 4, 9, 14, 24]

mocap = mocapLoader.load_mocap("./example_mocaps/walking_poses.npz")
mocap_keypoints = forwardKinematics.compute_fk(mocap.rots, apply_root_rotation=True)[FRAMES_TO_VISUALIZE]

color_list1 = [("#758bfd", "#758bfd") for _ in range(len(FRAMES_TO_VISUALIZE))]
color_list2 = [("#ff8600", "#ff8600") for _ in range(len(FRAMES_TO_VISUALIZE))]
color_list3 = [("#cccccc", "#cccccc") for _ in range(len(FRAMES_TO_VISUALIZE))]
color_list4 = [("#ffba08", "#efaa00") for _ in range(len(FRAMES_TO_VISUALIZE))]

plot_static_seq(
    poses_list=np.array([mocap_keypoints, mocap_keypoints]),
    body_model_type=SmplConstants.BODY_MODEL_TYPE,
    offset_next_pose=1,
    offset_sequences=2,
    rotation_axis=[0, 0, 1],
    rotation_angle=0,
    elevation=0,
    azimuth=-90,
    color_list=color_list1 + color_list3,
    link_size=1,
    keypoint_size=5,
    # save_svg_path='plot.svg',
    # show_grid=True,
    # fig_opts = {"figsize": (30, 30), "dpi": 80}
)
```
